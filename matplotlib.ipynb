```python
import numpy as np
import numpy as pandas
import matplotlib.pyplot as plt
```

```python
values=[10,20,30,40,50]
plt.plot(values)
plt.show()
```

```python
## Pie chart:- To see percentile distribution of different categories in the form of slices.
## 2 types of features:1. is numerical and 2. categorical.
```

```python
smedia=["Whatsapp","facebook","Instagram","telegram"]
suser=[200,250,500,150]
plt.pie(suser,labels=smedia,autopct="%0.2f",explode=[0,0,0,1])
plt.show()
```

```python
## Bar plot:- to see the relationship numerial and categorical features in bivarient analysis
```

```python
plt.bar(smedia,suser,color="purple")
plt.title("social media users",fontsize=20,fontweight="bold",pad=10)
plt.xlabel("Name of social media")
plt.ylabel(" USer")
plt.show()
```

```python
plt.barh(smedia,suser,color="blue")
plt.title("social media users",fontsize=20,fontweight="bold",pad=10)
plt.xlabel("Name of social media")
plt.ylabel(" USer")
plt.show()
```

```python
## Subplots
plt.subplot(2,2,1)
plt.subplot(2,2,2)
plt.subplot(2,2,3)
plt.subplot(2,2,4)
plt.show()
```

```python
plt.subplot(2,2,1)
values=[10,20,30,40,50]
plt.plot(values)

plt.subplot(2,2,2)
plt.boxplot(values)

plt.subplot(2,2,3)
plt.bar(smedia,suser,color="purple",width=0.40)

plt.subplot(2,2,4)
plt.pie(suser,labels=smedia,autopct="%0.2f")
plt.show()
```

## Stacked Bar chart


```python
country=["India","USA","Japan","Frans","China"]
gold=np.random.randint(1,16,(5,))
silver=np.random.randint(50,60,(5,))
bronze=np.random.randint(50,65,(5,))
```

```python
plt.bar(country,gold,color="yellow",label="gold")
plt.bar(country,silver,bottom=gold,color="red",label="silver")
plt.bar(country,bronze,bottom=silver,color="blue",label="Bronze")
plt.title("Medal count by country")
plt.xlabel("Country")
plt.ylabel("Medals")
plt.legend()
plt.show()
```
